```python
import pandas as pd
import sqlite3
```

```python
propre = pd.read_csv(r"C:\Users\mvond\Downloads\Apprentissage_Claude\donnees\accidents_propre.csv",
                     encoding="utf-8-sig", low_memory=False)
propre.shape
```

```python
# Création base de données et rangement
connexion = sqlite3.connect(r"C:\Users\mvond\Downloads\Apprentissage_Claude\donnees\accidents.db")
propre.to_sql("accidents", connexion, if_exists="replace", index=False)
connexion.close()
```

## Commentaire code
connect(...) ouvre la base. Si le fichier n’existe pas, il est créé.
to_sql("accidents", ...) copie ton tableau dans une table SQL nommée accidents.
if_exists="replace" remplace la table si elle existe déjà, ce qui te permet de relancer la cellule sans problème.
close() ferme la connexion proprement.
